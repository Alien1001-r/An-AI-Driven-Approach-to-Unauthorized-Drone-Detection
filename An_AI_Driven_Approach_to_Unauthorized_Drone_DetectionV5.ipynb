```python
# This code reads all XML annotation files from the DUT Anti-UAV training folder
# and lists all unique XML tag names to understand the dataset structure.

# Import necessary libraries
import os                              # Used to interact with files and folders
import xml.etree.ElementTree as ET     # Used to parse and read XML files
import pandas as pd                    # Used for data manipulation and creating DataFrames

# Display settings for pandas — show all columns and full width
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)

# ------------------------------
# Path to the training XML files
# ------------------------------
train_xml_path = "/content/drive/MyDrive/DUT_Anti_UAV/train/train/xml"

# ------------------------------
# Function to extract all unique XML tag names
# ------------------------------
def get_xml_columns(xml_folder):
    columns = set()  # Store unique tag names
    if os.path.exists(xml_folder):  # Check if path exists
        # Get all .xml files in the folder
        files = [f for f in os.listdir(xml_folder) if f.endswith('.xml')]
        for file in files:
            # Parse each XML file
            tree = ET.parse(os.path.join(xml_folder, file))
            root = tree.getroot()
            # Iterate through all XML elements and collect tag names
            for elem in root.iter():
                columns.add(elem.tag)
    return list(columns)

# ------------------------------
# Extract and display tag names for training data
# ------------------------------
train_columns = get_xml_columns(train_xml_path)
df_train_columns = pd.DataFrame([train_columns])

print(" Unique XML Tags in DUT Anti-UAV Training Dataset:")
display(df_train_columns)
```

####**3) DUT Anti-UAV XML to DataFrame Pipeline**

---


Converts XML annotations from the DUT Anti-UAV dataset into a structured DataFrame, where each row represents a single object. The DataFrame is ready for analysis or machine learning tasks.


```python
# ============================================================
# Full DUT Anti-UAV XML -> DataFrame pipeline (Step A-C, E-F)
# - Extract unique XML tags (structure check)
# - Parse all annotations into a DataFrame (actual data columns)
# - Step D (numeric conversion) has been separated
# - Save final CSV for later use
# ============================================================

import os
import xml.etree.ElementTree as ET
import pandas as pd

# ------------------------------
# Configuration: path to XMLs
# ------------------------------
xml_folder = "/content/drive/MyDrive/DUT_Anti_UAV/train/train/xml"

# Safety check
if not os.path.exists(xml_folder):
    raise FileNotFoundError(f"XML folder not found: {xml_folder}")

# -------------------------------------------------
# Step A — Inspect unique XML tags
# -------------------------------------------------
def get_unique_tags(folder, sample_limit=None):
    tags = set()
    files = [f for f in os.listdir(folder) if f.endswith('.xml')]
    if sample_limit:
        files = files[:sample_limit]
    for fname in files:
        tree = ET.parse(os.path.join(folder, fname))
        root = tree.getroot()
        for elem in root.iter():
            tags.add(elem.tag)
    return sorted(tags)

unique_tags = get_unique_tags(xml_folder, sample_limit=20)
print(" Unique XML Tags (example sample):")
print(unique_tags)
print(f" Total unique tags (sample): {len(unique_tags)}\n")

# -------------------------------------------------
# Step B — Parse all XML files into a record list
# -------------------------------------------------
records = []
xml_files = [f for f in os.listdir(xml_folder) if f.endswith('.xml')]
print(f"Found {len(xml_files)} XML files. Parsing all...")

for xml_file in xml_files:
    tree = ET.parse(os.path.join(xml_folder, xml_file))
    root = tree.getroot()

    # Image-level fields
    filename = root.findtext('filename')
    folder   = root.findtext('folder')
    path     = root.findtext('path')
    source_db = root.findtext('source/database')
    segmented = root.findtext('segmented')
    width_text  = root.findtext('size/width')
    height_text = root.findtext('size/height')
    depth_text  = root.findtext('size/depth')

    # Object-level fields
    for obj in root.findall('object'):
        obj_name   = obj.findtext('name')
        pose       = obj.findtext('pose')
        truncated  = obj.findtext('truncated')
        difficult  = obj.findtext('difficult')
        xmin_text  = obj.findtext('bndbox/xmin')
        ymin_text  = obj.findtext('bndbox/ymin')
        xmax_text  = obj.findtext('bndbox/xmax')
        ymax_text  = obj.findtext('bndbox/ymax')

        records.append({
            'filename': filename,
            'folder': folder,
            'path': path,
            'source': source_db,
            'database': source_db,
            'segmented': segmented,
            'width': width_text,
            'height': height_text,
            'depth': depth_text,
            'object_name': obj_name,
            'pose': pose,
            'truncated': truncated,
            'difficult': difficult,
            'xmin': xmin_text,
            'ymin': ymin_text,
            'xmax': xmax_text,
            'ymax': ymax_text
        })

# -------------------------------------------------
# Step C — Convert to DataFrame
# -------------------------------------------------
df_train_full = pd.DataFrame(records)

print("\n Parsed DataFrame summary:")
print(f"Rows (object annotations): {len(df_train_full)}")
print(f"Columns: {len(df_train_full.columns)}")
print("Column names:", df_train_full.columns.tolist())

# -------------------------------------------------
# Step E — Basic sanity checks and info
# -------------------------------------------------
#print("\nℹ Example data (first 5 rows):")
#print(df_train_full.head())

# -------------------------------------------------
# Step F — Save DataFrame to CSV
# -------------------------------------------------
output_csv = "/content/drive/MyDrive/DUT_Anti_UAV/train_annotations_full.csv"
df_train_full.to_csv(output_csv, index=False)
print(f"\n💾 CSV saved at: {output_csv}")
```

1. **XML Structure:** 21 unique tags, including `annotation`, `object`, `bndbox`, `filename`, `size`, `xmin/ymin/xmax/ymax`. Standard PASCAL VOC style.

2. **Dataset Size:** 5200 XML files, 5243 object annotations. Most images have 1 object.

3. **DataFrame:** 17 columns combining image-level and object-level info. Numeric fields ready for bbox computation.

```python
# List image and XML files
image_files = [f for f in os.listdir(src_images_folder) if f.lower().endswith((".jpg", ".jpeg", ".png"))]
xml_files = [f for f in os.listdir(src_labels_folder) if f.lower().endswith(".xml")]

# Match files by base name
image_basenames = [os.path.splitext(f)[0] for f in image_files]
xml_basenames = [os.path.splitext(f)[0] for f in xml_files]
common_basenames = list(set(image_basenames) & set(xml_basenames))

# Split into training and validation sets (80/20)
train_files, val_files = train_test_split(common_basenames, test_size=0.2, random_state=42)

print("Image–XML matching and dataset split completed successfully.")
```

##**Step 3 : Processing and converting XML annotations to YOLO format (TXT) :**


---
* In this step, we copied the image files to their respective training or validation folders and converted the corresponding XML annotations into YOLO TXT format.


```python
def process_and_copy_files(file_list, src_img_folder, src_xml_folder, dst_img_folder, dst_label_folder, classes):
    for filename in file_list:
        # Copy image file
        for ext in [".jpg", ".jpeg", ".png"]:
            src_img_path = os.path.join(src_img_folder, filename + ext)
            if os.path.exists(src_img_path):
                shutil.copy(src_img_path, os.path.join(dst_img_folder, filename + ext))
                break

        # Convert XML to YOLO TXT format
        src_xml_path = os.path.join(src_xml_folder, filename + ".xml")
        dst_txt_path = os.path.join(dst_label_folder, filename + ".txt")

        if os.path.exists(src_xml_path):
            tree = ET.parse(src_xml_path)
            root = tree.getroot()
            img_width = int(root.find("size/width").text)
            img_height = int(root.find("size/height").text)

            with open(dst_txt_path, "w") as f:
                for obj in root.findall("object"):
                    cls_name = obj.find("name").text
                    if cls_name not in classes:
                        continue
                    cls_id = classes.index(cls_name)

                    bbox = obj.find("bndbox")
                    xmin = int(bbox.find("xmin").text)
                    ymin = int(bbox.find("ymin").text)
                    xmax = int(bbox.find("xmax").text)
                    ymax = int(bbox.find("ymax").text)

                    # Normalize coordinates for YOLO format
                    x_center = ((xmin + xmax) / 2) / img_width
                    y_center = ((ymin + ymax) / 2) / img_height
                    width = (xmax - xmin) / img_width
                    height = (ymax - ymin) / img_height

                    f.write(f"{cls_id} {x_center:.6f} {y_center:.6f} {width:.6f} {height:.6f}\n")
```

###**Observation :**
* The function successfully processes each file by copying the corresponding image to the destination folder and converting its XML annotation to YOLO TXT format. For each file it ensures that only the specified classes are included normalizes the bounding box coordinates and prints a confirmation message, providing clear feedback on the processing status of each file.

```python
# Define the list of object classes/labels present in the dataset. This is essential for annotation formatting.
classes = ["UAV"]

# Print a status message indicating the start of processing and copying the training subset.
print("Processing and copying training files...")
process_and_copy_files(train_files, src_images_folder, src_labels_folder, dst_images_train, dst_labels_train, classes)
# Call a custom function to handle the training data:
# 1. Read files listed in 'train_files' (image and label names).
# 2. Convert label formats (e.g., from XML/CSV to YOLO/TXT format).
# 3. Copy the source images and new labels from their original folders to the designated training folders.

# Print a status message indicating the start of processing and copying the validation subset.
print("Processing and copying validation files...")
process_and_copy_files(val_files, src_images_folder, src_labels_folder, dst_images_val, dst_labels_val, classes)
# Call the custom function again to handle the validation data:
# 1. Read files listed in 'val_files'.
# 2. Convert and process their associated labels.
# 3. Copy the files to the designated validation folders.

# Print a final confirmation message upon successful completion of the data preparation and splitting process.
print(" Dataset split, processed, and copied successfully.")
```

#**Step 2 : Label Conversion**
* This step is important because it converts the annotation files (which describe the location of objects in each image) into a format that the CNN can understand and use during training. Without converting these labels, the model would not know where the object is or how to learn to detect it.

```python
import os                           # Provides functions for interacting with the operating system
import xml.etree.ElementTree as ET  # Used to parse and read XML files

# Folder that contains XML annotation files
xml_folder = "/content/drive/MyDrive/DUT_Anti_UAV/train/train/xml"

# Folder where YOLO formatted TXT labels will be saved
output_folder = "/content/drive/MyDrive/resnet_dataset/labels"

# Folder containing the training images (not used in conversion but kept for reference)
images_folder = "/content/drive/MyDrive/DUT_Anti_UAV/train/img"

# Create the output folder if it does not already exist
os.makedirs(output_folder, exist_ok=True)

# Function that converts VOC bounding box format to YOLO normalized format
def convert(size, box):
    dw = 1.0 / size[0]              # Normalization factor for width
    dh = 1.0 / size[1]              # Normalization factor for height

    x = (box[0] + box[2]) / 2.0     # Calculate x center
    y = (box[1] + box[3]) / 2.0     # Calculate y center

    w = box[2] - box[0]           # Bounding box width
    h = box[3] - box[1]           # Bounding box height

    return (x*dw, y*dh, w*dw, h*dh)  # Return normalized YOLO (x,y,w,h)

# Counter for number of files converted
count = 0

# Loop through all files in XML folder
for file in os.listdir(xml_folder):

    if not file.endswith(".xml"):  # Skip files that are not XML
        continue

    tree = ET.parse(os.path.join(xml_folder, file))   # Parse XML file
    root = tree.getroot()                           # Get root element of XML

    width = int(root.find("size/width").text)       # Read image width
    height = int(root.find("size/height").text)     # Read image height

    # Open a new TXT file to write YOLO labels
    out = open(os.path.join(output_folder, file.replace(".xml", ".txt")), "w")

    # Loop through every object (bounding box) in XML
    for obj in root.iter("object"):

        cls = 0   # Class ID (0 because only one class exists)

        xmlbox = obj.find("bndbox")  # Get bounding box data

        # Read xmin, ymin, xmax, ymax values
        b = (
            float(xmlbox.find("xmin").text),
            float(xmlbox.find("ymin").text),
            float(xmlbox.find("xmax").text),
            float(xmlbox.find("ymax").text)
        )

        bb = convert((width, height), b)  # Convert to YOLO format

        # Write class and bounding box to TXT file
        out.write(f"{cls} {' '.join([str(a) for a in bb])}\n")

    out.close()    # Close TXT file after writing
    count += 1     # Increase number of converted files

# Print summary of conversion
print(f"✅ Converted {count} XML files to YOLO TXT inside resnet_dataset/labels")
```

* A total of 5200 annotation files are now stored inside the resnet_dataset/labels folder.


#**Step 3 : Automatically Separating Authorized vs Unauthorized Images**


```python
import os
import shutil
import math
import random

# Define dataset paths for images and labels
images_dir = "/content/drive/MyDrive/resnet_dataset/images"
labels_dir = "/content/drive/MyDrive/resnet_dataset/labels"
auth_dir = "/content/drive/MyDrive/resnet_dataset/authorized"
unauth_dir = "/content/drive/MyDrive/resnet_dataset/unauthorized"

# Create output folders if they do not exist
os.makedirs(auth_dir, exist_ok=True)
os.makedirs(unauth_dir, exist_ok=True)

# Step 1 Collect information about bounding boxes
areas = []
file_info = []

# Loop through all label files
for file in os.listdir(labels_dir):

    # Ignore non txt files
    if not file.endswith(".txt"):
        continue

    # Build full path for label and image
    label_path = os.path.join(labels_dir, file)
    img_name = file.replace(".txt", ".jpg")
    img_path = os.path.join(images_dir, img_name)

    # Skip if image missing
    if not os.path.exists(img_path):
        continue

    # Read bounding box data
    with open(label_path, "r") as f:
        line = f.readline().strip()
        if not line:
            continue

        parts = line.split()
        if len(parts) != 5:
            continue

        # Extract bounding box values
        _, xc, yc, w, h = map(float, parts)

        # Compute bounding box area
        area = w * h
        # Compute distance from image center
        dist = math.sqrt((xc - 0.5)**2 + (yc - 0.5)**2)

        areas.append(area)
        file_info.append((img_path, area, dist))

# Step 2 Compute threshold values
if len(areas) == 0:
    print("No valid labels found!")
else:

    # Sort areas to find percentile
    areas_sorted = sorted(areas)
    area_threshold = areas_sorted[int(len(areas_sorted)*0.65)]
    dist_threshold = 0.5

    print(f" Auto thresholds set -> Area: {area_threshold:.6f}, Distance: {dist_threshold}")

    # Step 3 Split images based on thresholds
    authorized = 0
    unauthorized = 0

    for img_path, area, dist in file_info:

        # Generate random number to avoid name conflict
        rand = random.randint(1000,9999)
        name, ext = os.path.splitext(os.path.basename(img_path))
        new_name = f"{name}_{rand}{ext}"

        # Check if image satisfies authorized conditions
        if area >= area_threshold and dist <= dist_threshold:
            shutil.copy(img_path, os.path.join(auth_dir, new_name))
            authorized += 1
        else:
            shutil.copy(img_path, os.path.join(unauth_dir, new_name))
            unauthorized += 1

    print(" Splitting done!")
    print(f" Authorized: {authorized}")
    print(f" Unauthorized: {unauthorized}")
```

###**Observation :**
The output shows that automatic thresholds were successfully calculated based on the bounding box area and its distance from the image center.
Using these thresholds, the dataset has been split into two classes:

* **Authorized images:** 1776
* **Unauthorized images:** 3421

#**Step 4 : Train–Test Data Splitting**

```python
import os
import shutil
import random

# Define paths for folders containing classified images
auth_dir = "/content/drive/MyDrive/resnet_dataset/authorized"
unauth_dir = "/content/drive/MyDrive/resnet_dataset/unauthorized"

# Define destination folders for train and test datasets
train_dir = "/content/drive/MyDrive/resnet_dataset/train"
test_dir  = "/content/drive/MyDrive/resnet_dataset/test"

# Create train and test folders with subfolders for each class
for folder in [train_dir, test_dir]:
    for subfolder in ["authorized", "unauthorized"]:
        os.makedirs(os.path.join(folder, subfolder), exist_ok=True)

# Function to split images into train and test datasets
def split_data(src_dir, train_dest, test_dest, train_ratio=0.8):

    # Read all image files from the source directory
    files = [f for f in os.listdir(src_dir) if f.endswith(".jpg")]

    # Shuffle the list to randomize the split
    random.shuffle(files)

    # Determine the index where to split train and test
    split_idx = int(len(files) * train_ratio)

    # Create the train part
    train_files = files[:split_idx]

    # Create the test part
    test_files  = files[split_idx:]

    # Copy images assigned to training set
    for f in train_files:
        shutil.copy(os.path.join(src_dir, f), os.path.join(train_dest, f))

    # Copy images assigned to testing set
    for f in test_files:
        shutil.copy(os.path.join(src_dir, f), os.path.join(test_dest, f))

# Split authorized images into train and test
split_data(
    auth_dir,
    os.path.join(train_dir, "authorized"),
    os.path.join(test_dir, "authorized")
)

# Split unauthorized images into train and test
split_data(
    unauth_dir,
    os.path.join(train_dir, "unauthorized"),
    os.path.join(test_dir, "unauthorized")
)

# Notify user that splitting finished
print(" Dataset split into train and test successfully!")
```

###**Observation :**
* The output confirms that the dataset has been successfully divided into training and testing subsets.

#**Step 2 : Training the CNN Model with ResNet50**

```python
# Import necessary libraries
import torch
import torch.nn as nn
from torch.utils.data import DataLoader
from torchvision import datasets, transforms, models
import os

# Hyperparameters
epochs = 50
imgsz = 224
batch_size = 8
optimizer_name = "SGD"
lr0 = 0.01
patience = 10
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

project = "/content/drive/MyDrive/ResNetRuns"
name = "ResNet50_UAV"
save_path = f"{project}/{name}.pth"

# Create output folder
os.makedirs(project, exist_ok=True)

# Dataset path
data_dir = "/content/drive/MyDrive/resnet_dataset/train"

# Transform
transform = transforms.Compose([
    transforms.Resize((imgsz, imgsz)),
    transforms.ToTensor()
])

# Load dataset
train_data = datasets.ImageFolder(data_dir, transform=transform)

#  FAST DataLoader Settings
train_loader = DataLoader(
    train_data,
    batch_size=batch_size,
    shuffle=True,
    num_workers=2,       # Faster loading in Colab
    pin_memory=True      # Boosts GPU transfer speed
)

print("Number of classes:", len(train_data.classes))
print("Classes:", train_data.classes)

# Load pretrained ResNet50
model = models.resnet50(weights=models.ResNet50_Weights.DEFAULT)

# Replace last layer
model.fc = nn.Linear(model.fc.in_features, 2)

criterion = nn.CrossEntropyLoss()

# Optimizer
if optimizer_name == "SGD":
    optimizer = torch.optim.SGD(model.parameters(), lr=lr0, momentum=0.9)
else:
    optimizer = torch.optim.Adam(model.parameters(), lr=lr0)

model = model.to(device)

# Early stopping
best_loss = float('inf')
no_improve = 0

print("\n Training Started...\n")

# Training Loop
for epoch in range(epochs):

    model.train()
    epoch_loss = 0
    correct = 0
    total = 0

    for images, labels in train_loader:

        images = images.to(device, non_blocking=True)
        labels = labels.to(device, non_blocking=True)

        optimizer.zero_grad()
        outputs = model(images)

        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        epoch_loss += loss.item()

        _, predicted = torch.max(outputs, 1)
        total += labels.size(0)
        correct += (predicted == labels).sum().item()

    avg_loss = epoch_loss / len(train_loader)
    acc = 100 * correct / total

    print(f"[{epoch+1}/{epochs}] Loss:{avg_loss:.4f}  |  Acc:{acc:.2f}%")

    # Save best model
    if avg_loss < best_loss:
        best_loss = avg_loss
        no_improve = 0
        torch.save(model.state_dict(), save_path)
        print("✔ Saved best model →", save_path)
    else:
        no_improve += 1
        if no_improve >= patience:
            print("\n Stopped early (no improvement).")
            break

print("\n Training Completed!")
```

###**Observation :**

* The training process shows a consistent decrease in loss and a steady increase in accuracy until it reaches very high performance.
* The model achieves about 98–99% accuracy, which indicates that it has learned to distinguish between authorized and unauthorized images very well.

```python
import torch
# Import PyTorch library for building and using neural networks

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
# Import evaluation metrics to measure model performance

import time
# Import time module to measure inference speed

from torchvision import models, transforms, datasets
# Import pretrained models, image transformations, and dataset utilities

from torch.utils.data import DataLoader
# Import DataLoader to load data in batches

# ===== SETTINGS =====
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
# Choose GPU if available otherwise use CPU

best_model_path = "/content/drive/MyDrive/ResNetRuns/ResNet50_UAV.pth"
# Path where the trained ResNet model is stored

test_dir = "/content/drive/MyDrive/resnet_dataset/test"
# Folder containing test dataset images

# ===== DATA LOADING =====
transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor()
])
# Define transformations to resize images and convert them to tensors

test_data = datasets.ImageFolder(test_dir, transform=transform)
# Load test images and assign class labels automatically

test_loader = DataLoader(
    test_data,
    batch_size=8,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)
# Create DataLoader to iterate through test data efficiently

# ===== MODEL LOADING =====
model = models.resnet50(weights=None)
# Create ResNet50 architecture without pretrained weights

model.fc = torch.nn.Linear(model.fc.in_features, 2)
# Replace final layer to classify two classes

model.load_state_dict(torch.load(best_model_path, map_location=device))
# Load the trained model weights

model.to(device)
# Move model to GPU or CPU

model.eval()
# Set model to evaluation mode

# ===== TEST FUNCTION =====
def evaluate_test(model, loader, device):
    # Function that evaluates model on test dataset

    all_preds  = []
    all_labels = []
    # Lists to store predictions and ground truth labels

    total_images = 0
    start = time.time()
    # Prepare to measure FPS

    with torch.no_grad():
        # Disable gradient calculations for faster evaluation
        for images, labels in loader:

            images = images.to(device)
            labels = labels.to(device)
            # Move batch to GPU or CPU

            outputs = model(images)
            # Run model prediction

            _, preds = torch.max(outputs, 1)
            # Get predicted class index

            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())
            # Store results in CPU memory

            total_images += images.size(0)
            # Count processed images

    end = time.time()
    # Stop timer

    # ===== METRICS =====
    acc  = accuracy_score(all_labels, all_preds) * 100
    prec = precision_score(all_labels, all_preds, average="binary") * 100
    rec  = recall_score(all_labels, all_preds, average="binary") * 100
    f1   = f1_score(all_labels, all_preds, average="binary") * 100
    fps  = total_images / (end - start)
    # Compute accuracy, precision, recall, F1 score and speed

    print("\n📊 TEST RESULTS")
    # Print results title

    print(f"Accuracy:  {acc:.2f}%")
    print(f"Precision: {prec:.2f}%")
    print(f"Recall:    {rec:.2f}%")
    print(f"F1 Score:  {f1:.2f}%")
    print(f"FPS:       {fps:.2f}")
    # Display evaluation metrics

# ===== RUN TEST =====
evaluate_test(model, test_loader, device)
# Run evaluation on test set
```

#**Step 4 : Evaluation of YOLO Detection and ResNet Classification :**

```python
import os
import random
import torch
import torch.nn.functional as F
from torchvision import transforms
from PIL import Image
import matplotlib.pyplot as plt
from ultralytics import YOLO

# -----------------------------
# Paths
# -----------------------------
resnet_model_path = "/content/drive/MyDrive/ResNet50_UAV.pth"
yolo_model_path   = "/content/drive/MyDrive/yolo_dataset/best.pt"
images_root       = "/content/drive/MyDrive/resnet_dataset/images"

# -----------------------------
# Load ResNet50 model
# -----------------------------
from torchvision import models
model = models.resnet50(weights=None)
model.fc = torch.nn.Linear(model.fc.in_features, 2)
model.load_state_dict(torch.load(resnet_model_path, map_location="cpu"))
model.eval()

# -----------------------------
# Load YOLO model
# -----------------------------
yolo_model = YOLO(yolo_model_path)

# -----------------------------
# Image transform for ResNet
# -----------------------------
transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor()
])

# -----------------------------
# Function: ResNet prediction
# -----------------------------
def predict_resnet(img_path):
    img = Image.open(img_path).convert("RGB")
    img_t = transform(img).unsqueeze(0)

    with torch.no_grad():
        outputs = model(img_t)
        probs = F.softmax(outputs, dim=1)
        conf, pred = torch.max(probs, 1)

    label = "Authorized" if pred.item() == 0 else "Unauthorized"
    return label, conf.item() * 100

# -----------------------------
# Step 1: Find 3 Authorized + 3 Unauthorized FAST
# -----------------------------
authorized_selected = []
unauthorized_selected = []

all_images = os.listdir(images_root)
random.shuffle(all_images)

for img_name in all_images:
    img_path = os.path.join(images_root, img_name)

    # YOLO detection (skip images with no detections)
    yolo_result = yolo_model(img_path, conf=0.05)[0]
    if len(yolo_result.boxes) == 0:
        continue  # skip ugly/no-detection images

    # ResNet classification
    label, conf = predict_resnet(img_path)

    if label == "Authorized" and len(authorized_selected) < 3:
        authorized_selected.append(img_name)

    if label == "Unauthorized" and len(unauthorized_selected) < 3:
        unauthorized_selected.append(img_name)

    if len(authorized_selected) == 3 and len(unauthorized_selected) == 3:
        break

selected_images = authorized_selected + unauthorized_selected

print("✅ Selected Authorized:", authorized_selected)
print("✅ Selected Unauthorized:", unauthorized_selected)

# -----------------------------
# Step 2: Run YOLO + ResNet and show results
# -----------------------------
for img_name in selected_images:
    img_path = os.path.join(images_root, img_name)

    # YOLO detection
    yolo_result = yolo_model(img_path, conf=0.05)[0]
    yolo_img = yolo_result.plot()

    # ResNet prediction
    resnet_label, resnet_conf = predict_resnet(img_path)

    # Show image
    plt.figure(figsize=(6, 6))
    plt.imshow(yolo_img)
    plt.axis("off")
    plt.title(f"ResNet: {resnet_label} ({resnet_conf:.2f}%)")
    plt.show()
```

###**Observation :**
* The output shows that the system is able to detect the UAV using YOLOv8n and classify it using ResNet50 in real time.
* This confirms that both detection and classification work together successfully
